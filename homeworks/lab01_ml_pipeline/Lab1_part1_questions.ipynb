## ex. 1

$$  
y = x^Tx,  \quad x \in \mathbb{R}^N
$$

$$
\frac{dy}{dx} = \color{red}{2x}
$$

### Вывод ответа:

$$
y = x^T x = \sum_{i=1}^{N} x_i^2,
$$

частные производные по каждому компоненту $ x_j $:
$$
\frac{\partial y}{\partial x_j} = \frac{\partial}{\partial x_j} \left( x_1^2 + x_2^2 +...+x_N^2 \right) = 2x_j
$$

запишем тогда градиент:
$$
\frac{dy}{dx} = \begin{bmatrix}
\frac{\partial y}{\partial x_1} \\
\frac{\partial y}{\partial x_2} \\
\vdots \\
\frac{\partial y}{\partial x_N}
\end{bmatrix}
= \begin{bmatrix}
2x_1 \\
2x_2 \\
\vdots \\
2x_N
\end{bmatrix}
= 2x.
$$


## ex. 2

$$ y = tr(AB) \quad A,B \in \mathbb{R}^{N \times N} $$

$$
\frac{dy}{dA} = \color{red}{B^T}
$$

### Вывод ответа.
<br>
Выпишем след матрицы $ AB $ через элементы:
$$
y = \operatorname{tr}(AB) = \sum_{i=1}^{N} (AB)_{ii} = \sum_{i=1}^{N} \sum_{j=1}^{N} A_{ij} B_{ji} ~ , $$
частная производная функции $y$ по элементу $ A_{kl} $:
$$
\frac{\partial y}{\partial A_{kl}} = \frac{\partial}{\partial A_{kl}} \left( \sum_{i,j} A_{ij} B_{ji} \right) = B_{lk}~ ,
$$
так как только слагаемое с $ i=k, j=l $ содержит $ A_{kl} $, и его производная равна $ B_{lk} $, а остальные слагаемые дают $0$.
Соберём градиент как матрицу частных производных размера $ N \times N $:
$$
\frac{dy}{dA} =
\begin{bmatrix}
\frac{\partial y}{\partial A_{11}} & \frac{\partial y}{\partial A_{12}} & \cdots & \frac{\partial y}{\partial A_{1N}} \\
\frac{\partial y}{\partial A_{21}} & \frac{\partial y}{\partial A_{22}} & \cdots & \frac{\partial y}{\partial A_{2N}} \\
\vdots & \vdots & \ddots & \vdots \\
\frac{\partial y}{\partial A_{N1}} & \frac{\partial y}{\partial A_{N2}} & \cdots & \frac{\partial y}{\partial A_{NN}} \\
\end{bmatrix}
=
\begin{bmatrix}
B_{11} & B_{21} & \cdots & B_{N1} \\
B_{12} & B_{22} & \cdots & B_{N2} \\
\vdots & \vdots & \ddots & \vdots \\
B_{1N} & B_{2N} & \cdots & B_{NN} \\
\end{bmatrix}
= B^T
$$

## ex. 3

$$  
y = x^TAc , \quad A\in \mathbb{R}^{N \times N}, x\in \mathbb{R}^{N}, c\in \mathbb{R}^{N}
$$

$$
\frac{dy}{dx} = \color{red}{A c}
$$

$$
\frac{dy}{dA} = \color{red}{x c^T}
$$

Hint for the latter (one of the ways): use *ex. 2* result and the fact
$$
tr(ABC) = tr (CAB)
$$

### Вывод ответа.

Найдем $\frac{dy}{dx}$.

Рассмотрим $ y = x^T (A c) $. Обозначим $ z = A c $ — это вектор размера $ N \times 1 $, не зависящий от $ x $. Тогда:

$$
y = x^T z = \sum_{i=1}^{N} x_i z_i ~,
$$

затем аналогично заданию ex. 1, запишем $\frac{\partial y}{\partial x_j} = z_j = (A c)_j$, и градиент будет иметь вид:

$$
\frac{dy}{dx} =
\begin{bmatrix}
\frac{\partial y}{\partial x_1} \\
\frac{\partial y}{\partial x_2} \\
\vdots \\
\frac{\partial y}{\partial x_N}
\end{bmatrix}
=
\begin{bmatrix}
(A c)_1 \\
(A c)_2 \\
\vdots \\
(A c)_N
\end{bmatrix}
= A c
$$

#### Теперь $\frac{dy}{dA}$.
Pаметим, что $ y = x^T A c $ — это скаляр (вектор-строка × матрица × вектор-столбец). <br>
Воспользуемся результатом из ex.2 и подсказкой $\operatorname{tr}(ABC) = \operatorname{tr}(CAB)$.<br>
Так как $ y = x^T A c $ - это скаляр, а скаляр равен своему следу, запишем:

$$
y = x^T A c = \operatorname{tr}(x^T A c) ~.
$$

Теперь применим подсказку:

$$
\operatorname{tr}(x^T A c) = \operatorname{tr}(c x^T A)
$$

Теперь у нас выражение вида $ \operatorname{tr}(B A) $, где $ B = c x^T $. Это совпадает с формой из ex.2: $ y = \operatorname{tr}(BA) $, где $ B = c x^T $.

Из ex.2 мы знаем:

$$
\frac{d}{dA} \operatorname{tr}(BA) = B^T
$$

Подставляем $ B = c x^T $:

$$
\frac{dy}{dA} = (c x^T)^T = x c^T
$$

Применено транспонирование произведения: $ (AB)^T = B^T A^T $, и $ c^T $ -строка, $ x $ - столбец.

### Вывод ответа через подход 1:

Используем ex.2 и свойство нормы Фробениуса $|| X ||_F^2 = tr(XX^T) $, т.е. квадрат Фробениусовой нормы можно записать через след:

$$
\| X - AS \|_F^2 = \operatorname{tr}\left( (X - AS)^T (X - AS) \right)
$$

и раскроем скобки:

$$
J = \operatorname{tr}\left( X^T X - X^T AS - S^T A^T X + S^T A^T A S \right).
$$

Теперь возьмём производную по матрице $ S $. Воспользуемся результатом из ex.2:  
$$
\frac{d}{dS} \operatorname{tr}(B S) = B^T, \quad \frac{d}{dS} \operatorname{tr}(S^T C) = C, \quad \frac{d}{dS} \operatorname{tr}(S^T D S) = (D + D^T) S
$$

Применим это к каждому слагаемому:

- $ \frac{d}{dS} \operatorname{tr}(X^T X) = 0 $ - не зависит от $ S $,
- $ \frac{d}{dS} \operatorname{tr}(-X^T A S) = - (X^T A)^T = -A^T X $,
- $ \frac{d}{dS} \operatorname{tr}(-S^T A^T X) = -A^T X $,
- $ \frac{d}{dS} \operatorname{tr}(S^T A^T A S) = (A^T A + (A^T A)^T) S = 2 A^T A S $ — так как $ A^T A $ симметрична.

Суммируем все части:

$$
\frac{dJ}{dS} = -A^T X - A^T X + 2 A^T A S = -2 A^T X + 2 A^T A S,
$$

затем выносим общий множитель:

$$
\frac{dJ}{dS} = 2 A^T (AS - X) ~.
$$
